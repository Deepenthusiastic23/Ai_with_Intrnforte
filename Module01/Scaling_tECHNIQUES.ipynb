Feature Scaling 

```python
import pandas as pd
import numpy as np 

#  Sample dataframe 
data =  pd.DataFrame({
  "Height": [150,160,165,180,170],
  "Height": [65,72,78,85,70]
})
```

```python
data
```

Standard Scaler

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import numpy as np 

Standard_scale = StandardScaler()

scaled_data = Standard_scale.fit_transform(data)

scaled_data


```

```python


scaled_df = pd.DataFrame(
    scaled_data,
    columns=["scaled_height", ]
)

scaled_df
```
