Encoding Techniques

```python
import pandas as pd 
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
```

```python
data = pd.DataFrame({
  'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red'],
  'Size' : ['S', "M", "L", "M", "XL"],
  'Gender' : ['Male', 'Female', 'Female', 'Male', 'Male'],
  'Purchased' :['Yes', 'No', 'Yes', 'No', 'Yes']
})
data
```

```python
# 1. Label Encoding for 'Color'
labe1_encoder = LabelEncoder()
data['Color_Label'] = labe1_encoder.fit_transform(data['Color'])
data[["Color", 'Color_Label']]
```

```python
# Apply OneHotEncoding
onehot_encoder = OneHotEncoder()
color_encoded = onehot_encoder.fit_transform(data [['Gender']])

# Convert sparse matrix to array 
color_df = pd.DataFrame(color_encoded.toarray(), columns=onehot_encoder.get_feature_names_out(['Gender']))

color_df
```

```python
#Oridinal Encoding
size_order = [['S', 'M', 'L', 'XL']]
Ordinal_ncoder = OrdinalEncoder(categories=size_order)
data['Size_ordinal'] = Ordinal_ncoder.fit_transform(data[['Size']])
data[['Size', 'Size_ordinal']]
```

```python
#  Binary Endcoding  for 'Purchased
data['Purchased_Binary'] = data['Purchased'].map({'Yes':1 , 'No': 0})
data[['Purchased', 'Purchased_Binary']]
```

```python
# Final Encoded Data
encoded_data = pd.concat([data.drop(columns=['Color', 'Size', 'Gender', 'Purchased']), color_df], axis=1)
encoded_data
```
