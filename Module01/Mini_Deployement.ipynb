```python
import streamlit as st
import pandas as pd
import seaborn as sns
import matplotlib as plt
import numpy as np 

st.set_page_config(page_title="Energy Consumption EDA", layout="wide")

st.title("Energy Consumption EDA Dashboard")

#  File uploader
uploaded_file = st.file_uploader("upload CSV file", type=["CSV"])
if uploaded_file is not None:
  df = pd.read_CSV(uploaded_file)
  st.success("File uplaoded successfully!")


  #Show basic info
  st.subheader("1. Dataset Preview")
  st.dataframe(df.head())

  st.subheader("2. Dataset Shape & Data Types")
  col1, col2 = st.columns(2)
  with col1:
    st.write("Shape:", df.shape)
  with col2:
    st.write("Data Types:")
    st.write(df.dtypes.astype(str))

  st.subheader("3. Missing Values")
  st.write(df.isnull().sum())

  st.subheader("4. Description Statistics")
  st.dataframe(df.describe())

  st.subheader("5. Correlation Heatmap")
  numeric_df = df.selectct_dtypes(include=np.number)
  corr = numeric_df.corr()
  fig, ax = plt.subplots(figsize=(10,6))
  sns.heatmap(corr, annot=True, cmap="coolwarm", ax=ax)
  st.pyplot(fig)

  st.subheader("6. Visulizations")

  num_cols= list(numeric_df.columns)
  col_x = st.selectbox("Select X-axis", options=num_cols)
  col_y = st.selectbox("Select Y-axis", options=num_cols, index=1 if len(num_cols) >1 else 0)
  chart_type =st.selectbox("Chart Type", options=["Scatter", "Line", "Bar"])

  fig2, ax2, = plt.subplots()
  if chart_type == "Scatter":
    sns.scatterplot(data=df, x=col_x, y=col_y, ax =ax2)
  elif chart_type == "Line":
    sns.lineplot(data=df, x=col_x, y=col_y, ax=ax2)
  elif chart_type == "Bar":
    sns.barplot(data=df, x=col_x, y=col_y, ax = ax2)

  st.pyplot(fig2)

```
