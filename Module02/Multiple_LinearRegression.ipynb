 <!-- one output using two or more input  -->

```python
from sklearn.linear_model import LinearRegression
#  features : [Area , Bedrooms, age]

X = [
  [1000, 2, 2],
  [1500, 3, 8],
  [4500, 5, 3],
  [2500, 4, 1]
  ]

# House Prices in lakhs
y = [40,30,50,60]

model = LinearRegression()


# Train the model 
model.fit(X, y)

# Predict price of a new house
prediction = model.predict([[1800,3, 4]])

print("Predicted Price:", prediction[0], "Lakhs")
```

#  polynomial regression 
used when the relationship between input and output is curved rather than the straight line

```python
from sklearn.linear_model import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X  = [[1], [2],[3],[4], [5]]
y = [30,40,55, 75, 110]

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

prediction = model.predict(poly.transform([[6]]))

print("Predicted Salary:", prediction[0])
```
